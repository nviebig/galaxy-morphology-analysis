```python
# Master dark

dark_for_science = []
for i in range(1,10):
    data = fits.getdata(datapath + '/Dark60-00'+ str(i) + '.fit', ext=0)
    dark_for_science.append(data)
for i in range(10,20):
    data = fits.getdata(datapath + '/Dark60-0'+ str(i) + '.fit', ext=0)
    dark_for_science.append(data)

master_dark_for_science = np.median(dark_for_science, axis=0)
```

```python
# Master flat

flats = []
for i in range(1,10):
    data = fits.getdata(datapath + '/Flats-00'+ str(i) + 'G.fit', ext=0)
    head = fits.getheader(datapath + '/Flats-00'+ str(i) + 'G.fit', ext=0)
    flats.append(data)

data = fits.getdata(datapath + '/Flats-010G.fit', ext=0)
flats.append(data)


dark_for_flats = []
for i in range(1,10):
    data = fits.getdata(datapath + '/Dark0_2-00'+ str(i) + '.fit', ext=0)
    head = fits.getheader(datapath + '/Dark0_2-00'+ str(i) + '.fit', ext=0)
    dark_for_flats.append(data)

data = fits.getdata(datapath + '/Dark0_2-010.fit', ext=0)
dark_for_flats.append(data)


flats_ = np.zeros_like(flats) * 1.0
for i in range(len(flats)):
    flats_[i] = flats[i] - dark_for_flats[i]
    
master_flat = np.median(flats_, axis=0)
print(master_flat)
master_flat_mean = np.mean(master_flat)
```

```python
science_frames = np.zeros_like(raws) * 1.0
for i in range(len(raws)): 
    science_frames[i] = (raws[i] - master_dark_for_science)/(master_flat / master_flat_mean)
```
